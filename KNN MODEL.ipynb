KNN_MODEL: KNN (K-Nearest Neighbors) is a machine learning algorithm that makes predictions by looking at the nearest data points.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegressionCV
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
```

```python
df = pd.read_csv("train.csv")
df
```

```python
df.columns
```

```python
x = df[['battery_power', 'blue', 'clock_speed', 'dual_sim', 'fc', 'four_g',
       'int_memory', 'm_dep', 'mobile_wt', 'n_cores', 'pc', 'px_height',
       'px_width', 'ram', 'sc_h', 'sc_w', 'talk_time', 'three_g',
       'touch_screen', 'wifi']]

y = df['price_range']
```

```python
x_train,x_test,y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
knn_model = KNeighborsClassifier(n_neighbors =5)
```

```python
knn_model.fit(x_train,y_train)
```

```python
y_pred = knn_model.predict(x_test)
```

```python
print(classification_report(y_test,y_test))
```

This means the model has perfectly predicted all classes with 100% accuracy, precision, recall, and F1-score, with no misclassifications.
